```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
from sklearn.datasets import make_classification
x,y=make_classification(n_samples=1000,n_features=10,n_informative=7,n_redundant=3,n_classes=3,random_state=42)
```

```python
df=pd.DataFrame(x)
df['target']=y
```

```python
df.head()
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.3,random_state=42)
```

```python
x_train.shape
```

```python
from sklearn.neighbors import KNeighborsClassifier
```

```python
knn=KNeighborsClassifier()
```

```python
from sklearn.model_selection import GridSearchCV
```

```python
param_grid = {
    "n_neighbors": [3, 5, 7, 9, 11, 15, 21],
    "weights": ["uniform", "distance"],
    "metric": ["euclidean", "manhattan", "minkowski"],
    "p": [1, 2],
    "algorithm": ["auto", "ball_tree", "kd_tree", "brute"],
    "leaf_size": [20, 30, 40, 50]
}
```

```python
grid=GridSearchCV(estimator=knn,param_grid=param_grid,cv=5,scoring="accuracy",n_jobs=-1)
```

```python
grid.fit(x_train,y_train)
```

```python
y_pred=grid.predict(x_test)
```

```python
y_pred
```

```python
grid.best_params_
```

```python
grid.best_score_
```

```python
from sklearn.metrics import *
```

```python
accuracy_score(y_test,y_pred)
```

```python
param_grid = {
    "n_neighbors": [5, 7, 9, 11, 15, 21],
    "weights": ["uniform", "distance"],
    "metric": ["euclidean", "manhattan", "minkowski"],
    "p": [1, 2],
    "algorithm": ["auto", "ball_tree", "kd_tree", "brute"],
    "leaf_size": [30, 40, 50]
}
```

```python
grid_r=GridSearchCV(estimator=knr,param_grid=param_grid,cv=5,n_jobs=-1,scoring="neg_mean_squared_error")
```

```python
grid_r.fit(x_train,y_train)
```

```python
grid_r.best_params_
```

```python
grid_r.best_score_
```

```python
y_pred=grid_r.predict(x_test)
```
