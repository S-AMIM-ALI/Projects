```python
import pandas as pd
import numpy as np
import seaborn as sns 
import matplotlib.pyplot as plt
```

```python
from sklearn.datasets import make_classification
x,y=make_classification(n_samples=1000,n_features=10,n_redundant=5,n_informative=5,n_classes=2,random_state=1)
```

```python
from sklearn.naive_bayes import GaussianNB
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.svm import SVC
```

```python
nb_clf=GaussianNB()
lg_clf=LogisticRegression()
dt=DecisionTreeClassifier()
svc=SVC()
```

```python
from sklearn.ensemble import VotingClassifier
ensemble_clf=VotingClassifier(estimators=[('decision tree',dt),('naive bayes',nb_clf),('logistic',lg_clf),('svc',svc)],voting='hard')
```

```python
ensemble_clf
```

```python
df=pd.DataFrame(x)
```

```python
df['target']=y
```

```python
df.head()
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(df.iloc[:,:-1],df.iloc[:,-1],test_size=0.2,random_state=42)
```

```python
x_train.shape
```

```python
y_train.shape
```

```python
ensemble_clf.fit(x_train,y_train)
```

```python
y_pred=ensemble_clf.predict(x_test)
```

```python
y_pred
```

```python
from sklearn.metrics import *
```

```python
accuracy_score(y_test,y_pred)
```

```python
fpr, tpr, thresholds = roc_curve(y_test, y_pred)
```

```python
auc = roc_auc_score(y_test, y_pred)

print(fpr)
print(tpr)
print(auc)
```
