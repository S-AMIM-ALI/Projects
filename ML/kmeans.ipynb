```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
from sklearn.datasets import make_blobs
x,y=make_blobs(n_samples=1000,n_features=2,centers=5,random_state=42)
```

```python
x.shape,y.shape
```

```python
plt.scatter(x[:,0],x[:,1])
```

```python
from sklearn.cluster import KMeans
```

```python
kmeans=KMeans(n_clusters=5)
```

```python
kmeans.fit(x)
```

```python
y_labels=kmeans.predict(x)
```

```python
y_labels
```

```python
plt.scatter(x[:,0], x[:, 1], c=y_labels)
```

```python
from sklearn.metrics import silhouette_score
silhouette_score(x, y_labels)
```

```python
kmeans.labels_
```

```python
kmeans.inertia_ #wcss
```

```python
wcss = []
for k in range(1, 10):
    kmeans = KMeans(n_clusters=k)
    kmeans.fit(x)
    wcss.append(kmeans.inertia_)
```

```python
wcss
```

```python
plt.plot(range(1, 10), wcss)
plt.xticks(range(1, 10))
plt.xlabel("No of cluster,k")
plt.ylabel("WCSS")
plt.show()
```

```python
#hyperparameter tuning
```

```python
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score
```

```python

best_score = -1
best_model = None
for k in range(2, 11):

    model = KMeans(
        n_clusters=k,
        init="k-means++",
        n_init="auto",
        max_iter=100,
        algorithm="elkan",
        random_state=42
    )
    labels = model.fit_predict(x)

    score = silhouette_score(x, labels)

    if score > best_score:
        best_score = score
        best_model = model

```

```python
print("Best K:", best_model.n_clusters)
print("Best Silhouette Score:", best_score)
```
