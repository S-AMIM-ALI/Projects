```python
import pandas as pd
import numpy as np
import seaborn as sns 
import matplotlib.pyplot as plt
```

```python
from sklearn.datasets import load_iris
data =  load_iris()
```

```python
X = data.data
y = data.target
```

```python
data=pd.DataFrame(X)
data['target']=y
```

```python
data.head()
```

```python
data.shape
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(data.iloc[:,:-1], data.iloc[:,-1], test_size=0.33, random_state=1)
```

```python

from sklearn.preprocessing import StandardScaler #mean=0, sigma is 1
scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
```

```python
from sklearn.decomposition import PCA
pca = PCA()
```

```python
X_train_p = pca.fit_transform(X_train)
```

```python
X_train_p
```

```python
X_test=scaler.transform(X_test)
```

```python
X_test_p = pca.transform(X_test)
```

```python
pca.components_ #the eigenvectors of the covariance/correlation matri
```

```python
pca.explained_variance_ #Eigenvalues
```

```python
pca.explained_variance_ratio_ #percentage of variance captured/um of eigenvalues
```

```python
pcas = PCA(n_components=0.95) #This tells PCA to automatically keep enough components to retain approximately 95% of the variance.
```

```python
x_train_pca=pcas.fit_transform(X_train)
```

```python
x_test_pca=pcas.transform(X_test)
```

```python
pcas.components_
```

```python
pcas.explained_variance_ #these are eigenvalues and these are pricipal components pc1,pc2
```

```python
pcas.explained_variance_ratio_
```
