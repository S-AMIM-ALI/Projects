```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
from sklearn.datasets import make_classification
```

```python
x,y=make_classification(n_samples=1000,n_features=10,n_redundant=2,n_informative=8,n_classes=3)
```

```python
df=pd.DataFrame(x)
```

```python
df['target']=y
```

```python
df.head()
```

```python
df.isnull().sum()
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train,x_test,y_train,y_test=train_test_split(df.iloc[:,:-1],df.iloc[:,-1],test_size=0.3,random_state=42)
```

```python
from sklearn.ensemble import AdaBoostClassifier
from sklearn.tree import DecisionTreeClassifier
```

```python
ada=AdaBoostClassifier(estimator=DecisionTreeClassifier(random_state=42),
    random_state=42)
```

```python
param_grid = {
    "n_estimators": [50, 100, 200],
    "learning_rate": [0.01, 0.1, 0.5, 1.0],
    "estimator__max_depth": [1, 2, 3]
}
```

```python
from sklearn.model_selection import GridSearchCV
```

```python
grid = GridSearchCV(
    estimator=ada,
    param_grid=param_grid,
    cv=5,
    scoring="accuracy",
    n_jobs=-1,verbose=1
)
```

```python
grid.fit(x_train,y_train)
```

```python
print(grid.best_params_)
print(grid.best_score_)
```

```python
from sklearn.datasets import make_regression
x,y=make_regression(n_samples=1000,n_features=10,noise=0.3,random_state=42)
```

```python
df=pd.DataFrame(x)
df['target']=y
```

```python
df.head()
```

```python
x_train,x_test,y_train,y_test=train_test_split(df.iloc[:,:-1],df.iloc[:,-1],test_size=0.3,random_state=42)
```

```python
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import AdaBoostRegressor

base_tree = DecisionTreeRegressor(random_state=42)

model = AdaBoostRegressor(
    estimator=base_tree,
    random_state=42
)
```

```python
from sklearn.model_selection import GridSearchCV

param_grid = {
    "n_estimators": [50, 100, 200],
    "learning_rate": [0.01, 0.1, 0.5, 1.0],
    "estimator__max_depth": [1, 2, 3, 5]
}
```

```python
grid = GridSearchCV(
    estimator=model,
    param_grid=param_grid,
    cv=5,
    scoring="r2",
    n_jobs=-1,
    verbose=1
)
```

```python
grid.fit(x_train, y_train)
```

```python
grid.best_params_
```

```python
grid.best_score_
```
